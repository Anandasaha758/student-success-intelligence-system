```python
financial_features = [
    "Debtor",
    "Tuition fees up to date",
    "Scholarship holder"
]

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

for ax, column in zip(axes, financial_features):

    financial_data = pd.crosstab(
        df["Target"],
        df[column],
        normalize="index"
    ) * 100

    sns.heatmap(
        financial_data,
        annot=True,
        fmt=".1f",
        cmap="YlGnBu",
        linewidths=0.5,
        ax=ax,
        cbar=False
    )

    ax.set_title(column, fontsize=12, fontweight="bold")
    ax.set_xlabel("Recorded value")
    ax.set_ylabel("Student outcome")

fig.suptitle(
    "Financial Indicators Across Student Outcomes",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
financial_summary = df.groupby("Target").agg(
    Total_Students=("Target", "count"),
    Debtors=("Debtor", "sum"),
    Fees_Up_To_Date=("Tuition fees up to date", "sum"),
    Scholarship_Holders=("Scholarship holder", "sum")
)

financial_summary["Debtor_Percentage"] = (
    financial_summary["Debtors"] /
    financial_summary["Total_Students"] * 100
)

financial_summary["Fees_Up_To_Date_Percentage"] = (
    financial_summary["Fees_Up_To_Date"] /
    financial_summary["Total_Students"] * 100
)

financial_summary["Scholarship_Percentage"] = (
    financial_summary["Scholarship_Holders"] /
    financial_summary["Total_Students"] * 100
)

display(financial_summary.round(2))
```
